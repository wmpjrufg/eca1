## Escolha do caso

**A = apoio simples; E = engaste.** O eixo x acompanha o menor vão. Uma borda `x = constante` tem comprimento ly.

| Tipo | x=0 | x=lx | y=0 | y=ly |
|:--:|:--:|:--:|:--:|:--:|
| 1 | A | A | A | A |
| 2A | A | A | E | A |
| 2B | E | A | A | A |
| 3 | E | A | E | A |
| 4A | A | A | E | E |
| 4B | E | E | A | A |
| 5A | E | A | E | E |
| 5B | E | E | E | A |
| 6 | E | E | E | E |

Cada resultado inclui o desenho dos vínculos. **Todos** gera os nove casos, em um único Excel com capa ilustrada e uma aba por tipo.


```python
# @title 2. Selecionar a laje e as relações de vãos
Tipo = "3" # @param ["1", "2A", "2B", "3", "4A", "4B", "5A", "5B", "6", "Todos"]
Lambda_inicial = 1.0 # @param {type:"number"}
Lambda_final = 2.0 # @param {type:"number"}
Passo = 0.05 # @param {type:"number"}
```

```python
# @title Opções de precisão (pode manter os valores iniciais)
Tolerancia_percentual = 0.1 # @param {type:"number"}
Ordem_maxima = 40 # @param [20, 24, 28, 32, 36, 40, 44, 48] {type:"raw"}
```

```python
# @title 3. Calcular as tabelas com verificação de convergência
Tabelas = gerar_no_notebook(
    tipos=Tipo, inicio=Lambda_inicial, fim=Lambda_final, passo=Passo,
    poisson=0.20, ordens=range(8, Ordem_maxima + 1, 4),
    tolerancia=Tolerancia_percentual / 100,
)
```

## Leitura dos coeficientes

$$m_x=\mu_x\frac{p l_x^2}{100},\quad m_y=\mu_y\frac{p l_x^2}{100},\quad |m_x^-|=\mu'_x\frac{p l_x^2}{100},\quad |m_y^-|=\mu'_y\frac{p l_x^2}{100}.$$

$$w_{max}=\alpha_w\frac{p l_x^4}{100 E h^3},\qquad D=\frac{Eh^3}{12(1-\nu^2)}.$$

Os negativos aparecem em módulo. Os positivos e a flecha são máximos procurados na superfície; as posições estão no Excel. Os mapas mostram um exemplo de relação de vãos de cada tipo.

A precisão é acompanhada por duas mudanças sucessivas entre ordens. Essa medida não constitui um limite rigoroso do erro absoluto. Uma comparação com tabelas publicadas deve usar os mesmos vínculos, Poisson e normalização.

**Reações:** as tabelas usam $\bar r=k\,p\,l_x/10$ para a reação média de cada borda. Só as reações de borda são exportadas; as forças concentradas de canto entram apenas na verificação do equilíbrio. Por isso, a soma das bordas supera a carga total, a favor das vigas.



```python
# @title 4. Mostrar tabelas, apoios e mapas
exibir_resultados(Tabelas, lambda_visualizacao=1.5)
```
